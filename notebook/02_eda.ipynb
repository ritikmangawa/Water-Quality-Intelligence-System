```python
# Step 24 - Correlation matrix

correlation = df[water_quality_cols].corr()

correlation
```

```python
# Step 25 - Correlation heatmap

plt.figure(figsize=(10, 8))

plt.imshow(
    correlation,
    cmap="coolwarm",
    interpolation="nearest"
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(water_quality_cols)),
    water_quality_cols,
    rotation=90
)

plt.yticks(
    range(len(water_quality_cols)),
    water_quality_cols
)

plt.title("Correlation Between Water Quality Parameters")

plt.tight_layout()
plt.show()
```

```python
# Step 26 - Correlation with Dissolved Oxygen

do_correlation = correlation["Dissolved Oxygen (mg/l)"].sort_values(
    ascending=False
)

print(do_correlation)
```

```python
# Step 27 - Correlation with BOD

bod_correlation = correlation[
    "Biochemical Oxygen Demand (mg/l)"
].sort_values(
    ascending=False
)

print(bod_correlation)
```

```python
# Step 29 - DO vs Temperature

plt.figure(figsize=(8, 6))

plt.scatter(
    df["Temperature (cel)"],
    df["Dissolved Oxygen (mg/l)"],
    alpha=0.3
)

plt.xlabel("Temperature (°C)")
plt.ylabel("Dissolved Oxygen (mg/l)")
plt.title("Relationship Between Temperature and Dissolved Oxygen")

plt.tight_layout()
plt.show()
```

```python
# Step 30 - DO vs pH

plt.figure(figsize=(8, 6))

plt.scatter(
    df["pH (ph units)"],
    df["Dissolved Oxygen (mg/l)"],
    alpha=0.3
)

plt.xlabel("pH")
plt.ylabel("Dissolved Oxygen (mg/l)")
plt.title("Relationship Between pH and Dissolved Oxygen")

plt.tight_layout()
plt.show()
```

```python
# Step 31 - Convert Date for EDA

df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

print(df["Date"].dtype)
print("Invalid dates:", df["Date"].isnull().sum())
```

```python
# Step 32 - Extract year and month for EDA

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month

print(df[["Date", "Year", "Month"]].head())
```

```python
# Step 33 - Number of observations by year

year_counts = df["Year"].value_counts().sort_index()

plt.figure(figsize=(12, 5))

plt.plot(
    year_counts.index,
    year_counts.values
)

plt.title("Water Quality Observations Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Observations")

plt.tight_layout()
plt.show()
```

```python
# Step 34 - Average Dissolved Oxygen by year

yearly_do = df.groupby("Year")["Dissolved Oxygen (mg/l)"].mean()

plt.figure(figsize=(12, 5))

plt.plot(
    yearly_do.index,
    yearly_do.values
)

plt.title("Average Dissolved Oxygen Over Time")
plt.xlabel("Year")
plt.ylabel("Average DO (mg/l)")

plt.tight_layout()
plt.show()
```

```python
# Step 35 - Average BOD by year

yearly_bod = df.groupby("Year")[
    "Biochemical Oxygen Demand (mg/l)"
].mean()

plt.figure(figsize=(12, 5))

plt.plot(
    yearly_bod.index,
    yearly_bod.values
)

plt.title("Average BOD Over Time")
plt.xlabel("Year")
plt.ylabel("Average BOD (mg/l)")

plt.tight_layout()
plt.show()
```

```python
# Step 36 - Average DO by month

monthly_do = df.groupby("Month")[
    "Dissolved Oxygen (mg/l)"
].mean()

plt.figure(figsize=(9, 5))

plt.plot(
    monthly_do.index,
    monthly_do.values,
    marker="o"
)

plt.title("Average Dissolved Oxygen by Month")
plt.xlabel("Month")
plt.ylabel("Average DO (mg/l)")
plt.xticks(range(1, 13))

plt.tight_layout()
plt.show()
```

```python
# Step 37 - Average DO by waterbody type

waterbody_do = (
    df.groupby("Waterbody Type")["Dissolved Oxygen (mg/l)"]
    .mean()
    .sort_values(ascending=False)
)

print(waterbody_do)
```

```python
plt.figure(figsize=(10, 6))

waterbody_do.plot(kind="bar")

plt.title("Average Dissolved Oxygen by Waterbody Type")
plt.xlabel("Waterbody Type")
plt.ylabel("Average DO (mg/l)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Step 38 - Average DO by country

country_do = (
    df.groupby("Country")["Dissolved Oxygen (mg/l)"]
    .mean()
    .sort_values(ascending=False)
)

print(country_do)
```

```python
plt.figure(figsize=(8, 5))

country_do.plot(kind="bar")

plt.title("Average Dissolved Oxygen by Country")
plt.xlabel("Country")
plt.ylabel("Average DO (mg/l)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Step 39 - Data quality summary

summary = pd.DataFrame({
    "Feature": water_quality_cols,
    "Missing Values": [df[col].isnull().sum() for col in water_quality_cols],
    "Negative Values": [
        (df[col] < 0).sum()
        for col in water_quality_cols
    ],
    "Minimum": [df[col].min() for col in water_quality_cols],
    "Maximum": [df[col].max() for col in water_quality_cols]
})

summary
```

```python
# Step 40 - Skewness of water quality parameters

skewness = df[water_quality_cols].skew().sort_values(
    ascending=False
)

print(skewness)
```

```python
# Step 41 - Mean vs median

important_cols = [
    "Ammonia (mg/l)",
    "Biochemical Oxygen Demand (mg/l)",
    "Dissolved Oxygen (mg/l)",
    "Nitrate (mg/l)"
]

comparison = pd.DataFrame({
    "Mean": df[important_cols].mean(),
    "Median": df[important_cols].median(),
    "Std": df[important_cols].std()
})

comparison
```
