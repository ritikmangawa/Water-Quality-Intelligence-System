```python
# Step 8 - Create a random 100k row sample

import numpy as np

sample_size = 100000
chunksize = 100000

# First count rows without loading the entire dataset
total_rows = 0

for chunk in pd.read_csv(file_path, chunksize=chunksize):
    total_rows += len(chunk)

print("Total rows:", total_rows)
```

```python
# Step 9 - Randomly sample 100,000 rows

sample_size = 100000
chunksize = 100000

sampled_chunks = []

# Use a slightly larger fraction so we definitely get
# more than 100,000 rows
fraction = (sample_size + 2000) / total_rows

for chunk in pd.read_csv(file_path, chunksize=chunksize):
    
    n = min(
        len(chunk),
        int(len(chunk) * fraction) + 1
    )
    
    sampled_chunk = chunk.sample(
        n=n,
        random_state=42
    )
    
    sampled_chunks.append(sampled_chunk)

# Combine sampled rows
df_100k = pd.concat(
    sampled_chunks,
    ignore_index=True
)

# Randomly select exactly 100,000 rows
df_100k = df_100k.sample(
    n=sample_size,
    random_state=42
).reset_index(drop=True)

print("Final rows:", len(df_100k))
print("Final columns:", len(df_100k.columns))
```

```python
print(df_100k.shape)
```

```python
# Step 10 - Save the 100k working dataset

output_path = "../processed_data/water_quality_100k.csv"

df_100k.to_csv(output_path, index=False)

print("Dataset saved successfully!")
print("Saved at:", output_path)
```

```python
# Step 11 - Load the saved 100k dataset

df = pd.read_csv(output_path)

print("Shape:", df.shape)
```

```python
# Step 12 - Check missing values

missing = df.isnull().sum()

print(missing)
```

```python
# Step 13 - Check duplicate rows

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
# Step 14 - Check categorical information

print("Countries:", df["Country"].nunique())
print("Waterbody Types:", df["Waterbody Type"].nunique())

print("\nCountry distribution:")
print(df["Country"].value_counts())

print("\nWaterbody Type distribution:")
print(df["Waterbody Type"].value_counts())
```

```python
# Step 15 - Check date range

dates = pd.to_datetime(
    df["Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

print("Earliest date:", dates.min())
print("Latest date:", dates.max())

print("Invalid dates:", dates.isnull().sum())
```
